## Imports

```python
import numpy as np
from ert.storage import open_storage
```

## Set storage and experiment name

```python
storage_path = "/Users/LGAZ/ert/poly_example/storage"
experiment_name = "ensemble_experiment_6"
```

## Availabel experiments and ensembles

```python
with open_storage(storage_path) as storage:
    [print(f"Experiment names: {x.name}") for x in storage.experiments]
    [print(f"Ensamble names: {x.name}") for x in storage.ensembles]
```

## Get parameters and responses

```python
with open_storage(storage_path, "r") as storage:
    experiment = storage.get_experiment_by_name(experiment_name)
    ensemble = experiment.get_ensemble_by_name(
    "ensemble"
    )
    realizations = ensemble.get_realization_list_with_responses()
    
    # Parameters
    groups = list(experiment.parameter_configuration.keys())
    param_arrays = [
        ensemble.load_parameters(param_group, realizations, transformed = True)
        for param_group in groups
    ]
    X = np.vstack(param_arrays)

    # Responses
    Y = [ensemble.load_responses(response_type, (realizations)) for response_type in experiment.response_info]
```
